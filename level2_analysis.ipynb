```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Dataset .csv")
df.head()
```

```python
# level 2
# task  1 -Restaurant Ratings

rating_count = df['Aggregate rating'].value_counts().sort_index()

print("Rating Distribution:")
print(rating_count)

print("\nAverage Rating:")
print(df['Aggregate rating'].mean())

rating_count.plot(kind='hist', bins=10)
plt.title('Distribution of Restaurant Ratings')
plt.xlabel('Aggregate Rating')
plt.ylabel('Frequency')
plt.show()
```

```python
#level 2
# task 2 Top Cuisine Combination

combo = df['Cuisines'].value_counts().head(10)

print("Top 10 Cuisine Combinations:")
print(combo)

combo.plot(kind='bar')
plt.title('Top 10 Cuisine Combinations')
plt.xlabel('Cuisine Combination')
plt.ylabel('Number of Restaurants')
plt.xticks(rotation=90)
plt.show()
```

```python
# level 2 task 3
# task 3 - Geographic Analysis

print("Latitude Statistics:")
print(df['Latitude'].describe())

print("\nLongitude Statistics:")
print(df['Longitude'].describe())

plt.scatter(df['Longitude'], df['Latitude'], alpha=0.5)
plt.title('Restaurant Locations')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```

```python
# level 2 task 4
# Restaurant Chains

chains = df['Restaurant Name'].value_counts()
top_chains = chains[chains > 1].head(10)

print("Top Restaurant Chains:")
print(top_chains)

top_chains.plot(kind='bar')
plt.title('Top Restaurant Chains')
plt.xlabel('Restaurant Name')
plt.ylabel('Number of Outlets')
plt.xticks(rotation=90)
plt.show()
```
